```python
# Load environment variables (ANTHROPIC_API_KEY) from .env
from dotenv import load_dotenv

load_dotenv()
```

```python
# Streaming: receive the reply token-by-token instead of waiting for the whole
# thing - useful for responsive UIs. Kept raw here: our chat() returns a full
# string, which doesn't fit a stream you iterate over.
from anthropic import Anthropic
from anthropic.types import MessageParam

messages = [MessageParam(
    role="user",
    content="If a LLM is a statistical model, what's the reason behind the hype of AGI? Answer as clear, concise and precise as possible.")]

client = Anthropic()

# stream() opens a live connection; text_stream yields the text pieces as they arrive.
with client.messages.stream(
    model="claude-sonnet-4-5",
    max_tokens=1000,
    messages=messages,
) as stream:
    for text in stream.text_stream:
        print(text, end="")
```
