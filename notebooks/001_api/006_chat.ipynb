````python
# Chat is the outcome of this folder: it bundles the defaults we kept 
# re-passing and owns its own message history - so 001-005 collapse into a
# few lines each. Below is the same work we did by hand, now via Chat.

from utils.chat import Chat
````

````python
# 001 (messages): multi-turn history is automatic - Chat tracks it for us.
chat = Chat()

print(chat.user("Define quantum computing in one sentence.").send())
print(f"\n{chat.user("Elaborate further in one sentence.").send()}")  # sees the first turn
````

````python
# 002 (system prompt): set once at construction; it applies to every turn.
chat = Chat(system="You are a patient math tutor. Guide step by step; never give the direct answer.")

print(chat.user("How do I solve 5x + 3 = 2?").send())
````

````python
# 003 (temperature): an instance default, or a per-send() override.
print(Chat(temperature=1.0).user("Generate a one sentence movie idea").send())

# Newer models (Sonnet 5, Opus 4.8/4.7, Fable 5) removed sampling params: a
# non-default temperature returns a 400. Pin an older model to override it.
# (The default temperature=1.0 above is still accepted on the default model.)
print(Chat(model="claude-sonnet-4-5").user("Generate a one sentence movie idea").send(temperature=0.0))
````

````python
# 005 (structured output): the prefill + stop trick is a single send() argument.
# prefill is an assistant-turn prefill, which newer models (Sonnet 5,
# Opus 4.8/4.7, Fable 5) reject with a 400. Pin an older model that allows it.
rule = (
    Chat(model="claude-sonnet-4-5")
    .user("Generate a very short event bridge rule as json")
    .send(prefill="```json", stop_sequences=["```"])
)
print(rule)  # pure JSON, ready for json.loads
````
